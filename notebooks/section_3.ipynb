```python
import re
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
import pandas as pd
```

```python
df=pd.read_csv('../datasets/sentiment_analysis.csv')
df
```

```python
df=df[['text']]
```

```python
df
```

##### Data Preprocessing:
- Clean the text data (remove stopwords, punctuation, and special characters).
- Convert the text into numerical format using TF-IDF or Word Embedding (e.g.,
Word2Vec).

```python
stop_words = set(stopwords.words("english"))

df["text"] = df["text"].fillna("").astype(str)
df["text"] = df["text"].apply(lambda x: re.sub(r"[^a-zA-Z0-9\s-]+", "", x))
df["text"] = df["text"].apply(lambda x: " ".join([w for w in x.split() if w.lower() not in stop_words]))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfv=TfidfVectorizer()
X=tfv.fit_transform(df['text'])
```
